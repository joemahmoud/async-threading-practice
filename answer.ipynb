```python
import time
```

```python
def run_sync_tasks(task , delay) :
    print(f"{task} started")
    time.sleep(delay)
    print(f"{task} finshed")
```

```python
start = time.time()
run_sync_tasks("task 1" , 1)
run_sync_tasks("task 2" , 1)
run_sync_tasks("task 3" , 2)
run_sync_tasks("task 4" , 2)
run_sync_tasks("task 5" , 1)
run_sync_tasks("task 6" , 2)
run_sync_tasks("task 7" , 1)
run_sync_tasks("task 8" , 3)
print(time.time() - start , "s")
```

```python
import asyncio
```

```python
async def async_seq(task , delay) :
    print(f"{task} started")
    await asyncio.sleep(delay)
    print(f"{task} finshed")
```

```python
async def run_async_seq() :
 delays = [ 1 , 1 , 3 , 2 , 1 , 4 , 1 , 3]
 start = time.time()
 for i  in range(1 , 9) :
       task = f"task {i}"
       await async_seq( task , delays[i-1])
 print(time.time() - start)

await run_async_seq()
```

```python
async def async_test(task , delay) :
    print(f"{task} started")
    await asyncio.sleep(delay)
    print(f"{task} finshed")
```

```python
async def run_async() :
     tasks = [async_test("task 1" , 2) , async_test("test 2" , 1) , async_test("test 3" , 3) , async_test("test 4" , 3)]
     start = time.time()
     await asyncio.gather(*tasks)
     print(time.time() - start)

await run_async()
```

```python
def heavy_cpu_function() :
    result = 1
    for i in range(1 , 1000) :
        result = result * i

    print(result)
```

```python
async def test() :
    print("start")
    for i in range(1 , 6) :
        if i == 3 :
         await asyncio.sleep(2)
        print(i)
        
    print("finshed")
    
```

```python
async def  main () :

    heavy_cpu_function()
    await test()
    
await main()
```

```python
async def counter_yield() :
    print("counter_yield started")

    for i in range(1 , 6) :
        print("A" , i)
        await asyncio.sleep(0)

    print("counter_yield finshed")


async def counter_yield_2() :
     print("counter_yield_2 started")
     for i in range(1 , 6) :
             print("B" , i)
             await asyncio.sleep(0)
     
     print("counter_yield_2 finshed")

```

```python
async def yield_run(): 
    await asyncio.gather(counter_yield() , counter_yield_2())



await yield_run()
```

```python
import threading
```

```python
shared_counter = 0 
shared_counter_2 = 0
lock = threading.Lock()
def race_condation() :
    print("race start")
    global shared_counter
    with lock :
      shared_counter += 1
      print(shared_counter)
    print("race finsh")


def race_condation_2() :
    print("race start")
    global shared_counter_2
    
    shared_counter_2 += 1
    print(shared_counter_2)
    print("race finsh")
```

```python
def run_race() :
    t1 = threading.Thread(target=race_condation)
    t2 = threading.Thread(target=race_condation)
    t1.start()
    t2.start()
    t1.join()
    t2.join()

def run_race_2() :
    t1 = threading.Thread(target=race_condation_2)
    t2 = threading.Thread(target=race_condation_2)
    t1.start()
    t2.start()
    t1.join()
    t2.join()



run_race()
run_race_2()
```
